## 4) Diagramas de flujo **[5 pts]**
- ### Diagrama 1: Flujo general
   - Se lee n y los n códigos. Se crean dos árboles vacíos (BST y AVL). Con un ciclo, cada código se inserta en ambos árboles, en el mismo orden. Al terminar, se dibujan los dos árboles, se recorren en preorden, inorden y postorden, y se muestran los resultados.


- ### Diagrama 2: Inserción en el BST (recursiva, sin balanceo)

    - Si el nodo es None, se crea uno nuevo ahí (caso base). Si el valor es menor que el nodo actual, se inserta a la izquierda; si es mayor, a la derecha; si es igual, se ignora (duplicado). La función siempre retorna el nodo, para reconstruir las conexiones al subir en la recursión. No hay reorganización: la forma del árbol depende solo del orden de llegada.



- ### Diagrama 3: Inserción en el AVL (recursiva, con balanceo)

   - Se inserta igual que en el BST. Luego, al regresar de la recursión, se actualiza la altura del nodo y se calcula el factor de balance fb = h_izq − h_der. Si |fb| ≤ 1, no se hace nada. Si |fb| > 1, se aplica una de cuatro rotaciones según el caso:

        - izq-izq → simple derecha
        - izq-der → doble izq-der
        - der-der → simple izquierda
        - der-izq → doble der-izq

Se retorna la (posible nueva) raíz del subárbol.



- ### Diagrama 4: Recorridos (preorden, inorden, postorden)

    - Si el nodo es None, se retorna. Si no, se ejecutan tres acciones —visitar la raíz, recorrer la izquierda, recorrer la derecha— pero en distinto orden:

        - Preorden: raíz, izquierda, derecha
        - Inorden: izquierda, raíz, derecha
        - Postorden: izquierda, derecha, raíz

En un BST, el inorden siempre entrega los valores ordenados de menor a mayor.



### 5) Análisis de complejidad **[15 pts]**
## Análisis de complejidad **[15 pts]**

### Justificación

El análisis de complejidad permite comparar el comportamiento del BST y el AVL a medida que aumenta la cantidad de códigos ingresados.

En el **BST**, la complejidad depende de la altura del árbol. Cuando el árbol está aproximadamente balanceado, las operaciones pueden realizarse en $O(\log n)$. Sin embargo, cuando los datos generan un árbol completamente inclinado, la altura puede llegar a $n$ y una inserción puede tener una complejidad de $O(n)$.

En el **AVL**, se utilizan rotaciones para mantener el árbol balanceado. Por esta razón, su altura se mantiene en $O(\log n)$ y la inserción conserva esta complejidad incluso cuando los datos llegan en un orden desfavorable.


### Función T(n)

#### BST

La complejidad de una inserción depende de la altura del árbol:

$$
T(n)=O(h)
$$

**Mejor caso:**

Cuando el árbol está aproximadamente balanceado:

$$
T(n)=T(n/2)+c
$$

Por lo tanto:

$$
\boxed{T(n)=O(\log n)}
$$

**Peor caso:**

Cuando el árbol queda completamente inclinado:

$$
T(n)=T(n-1)+c
$$

Por lo tanto:

$$
\boxed{T(n)=O(n)}
$$


#### AVL

El AVL mantiene una altura logarítmica mediante el balanceo y las rotaciones:

$$
T(n)=T(n/2)+c
$$

Por lo tanto:

$$
\boxed{T(n)=O(\log n)}
$$

Las rotaciones simples y dobles tienen un costo constante:

$$
O(1)
$$


### Mejor y peor caso

| Estructura | Mejor caso | Peor caso |
|---|---:|---:|
| **BST** | $O(\log n)$ | $O(n)$ |
| **AVL** | $O(\log n)$ | $O(\log n)$ |


### Justificación de los resultados

El BST no realiza ningún balanceo, por lo que su comportamiento depende del orden en que llegan los códigos. Si los datos generan una estructura muy inclinada, el árbol puede comportarse de forma similar a una lista.

El AVL calcula el factor de balance y realiza rotaciones cuando es necesario. Esto evita que el árbol alcance una altura lineal y permite mantener una altura logarítmica.

Por esta razón, el BST puede alcanzar $O(n)$ por inserción en el peor caso, mientras que el AVL mantiene $O(\log n)$.


### Gráfica de mejor y peor caso

La gráfica compara:

- **BST mejor caso:** $O(\log n)$
- **BST peor caso:** $O(n)$
- **AVL:** $O(\log n)$

La gráfica permite observar que el peor caso del BST crece más rápidamente que los casos logarítmicos.

```python
import matplotlib.pyplot as plt

''' node structure'''
class node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        self.height = 1

''' BST fuctions '''
def insert_bst(root, value):
    if root is None:
        return node(value)
    if value < root.value:
        root.left = insert_bst(root.left, value)
    elif value > root.value:
        root.right = insert_bst(root.right, value)
    return root

''' AVL tree functions '''
def obtain_height(node):
    if not node:
        return 0
    return node.height

def obtain_balance(node):
    if not node:
        return 0
    return obtain_height(node.left) - obtain_height(node.right)

def right_rotation(y):
    x = y.left
    T2 = x.right
    x.right = y
    y.left = T2
    y.height = 1 + max(obtain_height(y.left), obtain_height(y.right))
    x.height = 1 + max(obtain_height(x.left), obtain_height(x.right))
    return x

def left_rotation(x):
    y = x.right
    T2 = y.left
    y.left = x
    x.right = T2
    x.height = 1 + max(obtain_height(x.left), obtain_height(x.right))
    y.height = 1 + max(obtain_height(y.left), obtain_height(y.right))
    return y

def insert_avl(root, value):
    if not root:
        return node(value)
    elif value < root.value:
        root.left = insert_avl(root.left, value)
    elif value > root.value:
        root.right = insert_avl(root.right, value)
    else:
        return root

    root.height = 1 + max(obtain_height(root.left), obtain_height(root.right))
    balance = obtain_balance(root)

    if balance > 1 and value < root.left.value:
        return right_rotation(root)
    if balance < -1 and value > root.right.value:
        return left_rotation(root)
    if balance > 1 and value > root.left.value:
        root.left = left_rotation(root.left)
        return right_rotation(root)
    if balance < -1 and value < root.right.value:
        root.right = right_rotation(root.right)
        return left_rotation(root)

    return root

''' traversals '''
def preorder(root, result):
    if root:
        result.append(root.value)
        preorder(root.left, result)
        preorder(root.right, result)

def inorder(root, result):
    if root:
        inorder(root.left, result)
        result.append(root.value)
        inorder(root.right, result)

def postorder(root, result):
    if root:
        postorder(root.left, result)
        postorder(root.right, result)
        result.append(root.value)

''' matplotlib functions '''
def build_coords(node, x, y, dx, dy, coords, connections):
    if node:
        coords[node.value] = (x, y)
        if node.left:
            connections.append((node.value, node.left.value))
            build_coords(node.left, x - dx, y - dy, dx / 2, dy, coords, connections)
        if node.right:
            connections.append((node.value, node.right.value))
            build_coords(node.right, x + dx, y - dy, dx / 2, dy, coords, connections)

def tree_graph(root, tittle):
    if not root:
        return
    coords = {}
    connections = []
    build_coords(root, 0, 0, 10, 2, coords, connections)

    fig, ax = plt.subplots(figsize=(10, 6))
    ax.set_title(tittle, fontsize=14, fontweight='bold')
    ax.axis('off')

    for father, son in connections:
        x_vals = [coords[father][0], coords[son][0]]
        y_vals = [coords[father][1], coords[son][1]]
        ax.plot(x_vals, y_vals, 'k-', lw=1.5, zorder=1)

    for value, (x, y) in coords.items():
        ax.plot(x, y, 'o', markersize=32, color="#DB63DD", markeredgecolor='black', zorder=2)
        ax.text(x, y, str(value), ha='center', va='center', fontsize=10, color='black', fontweight='bold', zorder=3)

    plt.show()

''' FUNCTIONS AND MENU '''
def ask_int_num(mensaje):
    while True:
        num = input(mensaje).strip()
        if num.isdigit():
            return int(num)
        print("\n\t\tPor favor ingrese un número válido.")

def main():
    print("\n\t=== ORGANIZACIÓN DE CÓDIGOS MEDIANTE ÁRBOLES ===\n")

    n = ask_int_num("\tIngrese la cantidad de productos a registrar: ")
    codes = []
    for i in range(1, n + 1):
        cod = ask_int_num(f"\tIngrese el código {i}: ")
        codes.append(cod)

    print(f"\n\tLista de códigos registrados: {codes}")

    '''Árboles'''
    root_bst = None
    root_avl = None

    for code in codes:
        root_bst = insert_bst(root_bst, code)
        root_avl = insert_avl(root_avl, code)

    while True:
        print("\n\t\t===== MENÚ DE OPERACIONES =====")
        print("\t1. Ver gráficas (Comparación BST vs AVL)")
        print("\t2. Recorrido Preorden")
        print("\t3. Recorrido Inorden (Análisis de ordenamiento)")
        print("\t4. Recorrido Postorden")
        print("\t0. Salir")
        option = input("\n\tSeleccione una opción: ").strip()

        if option == "1":
            print("\n\t\tGenerando gráficas en ventanas externas...")
            tree_graph(root_bst, "Árbol Binario de Búsqueda (BST) - Sin Balancear")
            tree_graph(root_avl, "Árbol AVL - Auto-Balanceado")

        elif option == "2":
            res_bst, res_avl = [], []
            preorder(root_bst, res_bst)
            preorder(root_avl, res_avl)
            print(f"\n\tPreorden BST: {res_bst}")
            print(f"\tPreorden AVL: {res_avl}")

        elif option == "3":
            res_bst, res_avl = [], []
            inorder(root_bst, res_bst)
            inorder(root_avl, res_avl)
            print(f"\n\tInorden BST: {res_bst}")
            print(f"\tInorden AVL: {res_avl}")
            print("\n\t[!] ANÁLISIS DEL RECORRIDO INORDEN:")
            print("\tAl aplicar Inorden sobre un Árbol Binario de Búsqueda (o AVL),")
            print("\tlos valores resultan estar perfectamente ordenados de menor a mayor.")
            print("\tEsto ocurre porque se lee primero la rama izquierda (menores),")
            print("\tluego la raíz, y finalmente la rama derecha (mayores).")

        elif option == "4":
            res_bst, res_avl = [], []
            postorder(root_bst, res_bst)
            postorder(root_avl, res_avl)
            print(f"\n\tPostorden BST: {res_bst}")
            print(f"\tPostorden AVL: {res_avl}")

        elif option == "0":
            print("\n\tCerrando el sistema...")
            break
        else:
            print("\n\tOpción inválida.")

if __name__ == "__main__":
    main()546
```
