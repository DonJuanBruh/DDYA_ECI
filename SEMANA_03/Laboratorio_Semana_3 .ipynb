## 8) Código funcional **[15 pts]**


```python
def keep_products_id(products):
    id_list_nomerge = []
    for product in range(products):
        print("\n\t\tProducto #", product + 1)
        while True:
            id = int(input("\tID/Código = "))
            if id > 0:
                break
            print("\n\t\tValor ingresado no válido. Por favor vuelva a ingresar un ID:\n")
        id_list_nomerge.append(id)
    return id_list_nomerge
 
def merge_sort(id_products_list):
    if len(id_products_list) <= 1:
        return id_products_list
 
    mid = len(id_products_list) // 2
    l_half = id_products_list[:mid]
    r_half = id_products_list[mid:]
 
    left_sorted = merge_sort(l_half)
    right_sorted = merge_sort(r_half)
 
    return _merge(left_sorted, right_sorted)
 
def _merge(left, right):
    sorted_list = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j +=1
    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])
 
    return sorted_list
 
def binary_search(products_id_sorted, id):
    left = 0
    right = len(products_id_sorted) - 1
 
    while left <= right:
        mid = (left + right) // 2
       
        if products_id_sorted[mid] == id:
            return mid
        elif products_id_sorted[mid] < id:
            left = mid + 1
        else:
            right = mid - 1
    return -1
 
def main_menu():
    print("\n\t\t--ORGANIZACIÓN Y BÚSQUEDA by FairSociety--")
    products = int(input("\nIngrese la cantidad de productos que desea registrar:"))
    id_products_list = keep_products_id(products)
    products_id_sorted = merge_sort(id_products_list)
    print("\n\tLa lista ordenada de los códigos/id de los productos es:", products_id_sorted)
    print("\n\tAhora podrá buscar el código/id de algún producto para comprobar si está registrado según la lista.")
    id = int(input("\n\t\tIngrese el código/id a buscar:"))
    result = binary_search(products_id_sorted, id)
 
    if result != -1:
        print(f"\n\tEl producto con código/id {id} sí está registrado.")
    else:
        print(f"\n\tEl producto con código/id {id} no está registrado.")
 
    print("\nFin del programa")
 
 
if __name__ == "__main__":
    main_menu()
```
